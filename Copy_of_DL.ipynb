Feature engineering
Feature-engineering cell containing groupby, shift, lag_1, lag_2, lag_3, previous_gap_sec and the historical feature calculations.

```python

# PHASE 4: TEMPORAL FEATURE ENGINEERING

import numpy as np
import pandas as pd

# Group by region and function to avoid mixing unrelated functions
group_cols = ['region', 'functionName']

# 1. Sort chronologically
df_model = df_clean.sort_values(
    group_cols + ['timestamp']
).reset_index(drop=True)

# 2. Previous cold-start latencies
grouped = df_model.groupby(group_cols, sort=False)

df_model['lag_1'] = grouped['latency'].shift(1)
df_model['lag_2'] = grouped['latency'].shift(2)
df_model['lag_3'] = grouped['latency'].shift(3)

# 3. Number of previous events
df_model['history_count'] = grouped.cumcount()

# 4. Previous three-event average
df_model['prev3_mean'] = (
    df_model.groupby(group_cols)['lag_1']
    .transform(lambda x: x.rolling(3, min_periods=1).mean())
)

# 5. Previous three-event standard deviation
df_model['prev3_std'] = (
    df_model.groupby(group_cols)['lag_1']
    .transform(lambda x: x.rolling(3, min_periods=2).std())
)

# 6. Time since previous recorded cold-start event
df_model['time_gap_sec'] = (
    df_model.groupby(group_cols)['timestamp']
    .diff()
    .dt.total_seconds()
)

# 7. Cyclical time features
hour_float = (
    df_model['hour'] +
    df_model['minute'] / 60
)

df_model['hour_sin'] = np.sin(
    2 * np.pi * hour_float / 24
)

df_model['hour_cos'] = np.cos(
    2 * np.pi * hour_float / 24
)

# 8. Remove first event of each function:
# It has no previous event available for prediction.
df_model = df_model[
    df_model['history_count'] > 0
].copy()

# 9. Fill unavailable older history with the most recent
# available previous latency.
df_model['lag_2'] = df_model['lag_2'].fillna(
    df_model['lag_1']
)

df_model['lag_3'] = df_model['lag_3'].fillna(
    df_model['lag_1']
)

df_model['prev3_std'] = df_model['prev3_std'].fillna(0)

# 10. Log transformations for skewed input features
df_model['log_time_gap'] = np.log1p(
    df_model['time_gap_sec'].clip(lower=0)
)

df_model['log_memory'] = np.log1p(
    df_model['memoryMB']
)

# 11. Inspect results
print("FEATURE-ENGINEERED DATASET")
print("Shape:", df_model.shape)

print("\nMissing values in engineered features:")
feature_cols = [
    'lag_1', 'lag_2', 'lag_3',
    'prev3_mean', 'prev3_std',
    'time_gap_sec', 'log_time_gap',
    'hour_sin', 'hour_cos',
    'log_memory', 'history_count'
]

print(df_model[feature_cols].isnull().sum())

print("\nFeature examples:")
display(
    df_model[
        [
            'functionName', 'region', 'runtime',
            'memoryMB', 'latency',
            'lag_1', 'lag_2', 'lag_3',
            'prev3_mean', 'prev3_std',
            'time_gap_sec'
        ]
    ].head(10)
)

print("\nHistory count distribution:")
print(df_model['history_count'].describe())
```

```python

# PHASE 5: GLOBAL CHRONOLOGICAL DATA SPLIT

# Sort all records by actual timestamp
df_model = df_model.sort_values(
    'timestamp'
).reset_index(drop=True)

# Determine chronological cutoff timestamps
time_70 = df_model['timestamp'].quantile(0.70)
time_85 = df_model['timestamp'].quantile(0.85)

# Split using timestamps, not random sampling
train_df = df_model[
    df_model['timestamp'] < time_70
].copy()

val_df = df_model[
    (df_model['timestamp'] >= time_70) &
    (df_model['timestamp'] < time_85)
].copy()

test_df = df_model[
    df_model['timestamp'] >= time_85
].copy()

# Display split information
print("TRAINING DATA")
print("Records:", len(train_df))
print("Start:", train_df['timestamp'].min())
print("End:", train_df['timestamp'].max())

print("\nVALIDATION DATA")
print("Records:", len(val_df))
print("Start:", val_df['timestamp'].min())
print("End:", val_df['timestamp'].max())

print("\nTEST DATA")
print("Records:", len(test_df))
print("Start:", test_df['timestamp'].min())
print("End:", test_df['timestamp'].max())

print("\nSPLIT PERCENTAGES")
total = len(df_model)

print("Train:", round(len(train_df) / total * 100, 2), "%")
print("Validation:", round(len(val_df) / total * 100, 2), "%")
print("Test:", round(len(test_df) / total * 100, 2), "%")

# Verify chronological separation
assert train_df['timestamp'].max() < val_df['timestamp'].min()
assert val_df['timestamp'].max() < test_df['timestamp'].min()

print("\nChronological split verified!")
```

```python

# PHASE 6A: LEAKAGE-SAFE MODEL PREPROCESSING

import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler

# ------------------------------------------------
# 1. Recalculate previous-event time information
# ------------------------------------------------

df_model = df_model.sort_values('timestamp').reset_index(drop=True)

grouped = df_model.groupby(
    ['region', 'functionName'],
    sort=False
)

# Timestamp of the previous event for this function
df_model['previous_timestamp'] = (
    grouped['timestamp'].shift(1)
)

# Gap known from the previous event, NOT the target event
df_model['previous_gap_sec'] = (
    grouped['time_gap_sec'].shift(1)
)

df_model['previous_gap_sec'] = (
    df_model['previous_gap_sec'].fillna(0).clip(lower=0)
)

# Previous event's time of day
previous_hour = df_model['previous_timestamp'].dt.hour.fillna(0)
previous_minute = df_model['previous_timestamp'].dt.minute.fillna(0)

previous_hour_float = previous_hour + previous_minute / 60

df_model['previous_hour_sin'] = np.sin(
    2 * np.pi * previous_hour_float / 24
)

df_model['previous_hour_cos'] = np.cos(
    2 * np.pi * previous_hour_float / 24
)

# ------------------------------------------------
# 2. Recreate chronological splits
# ------------------------------------------------

train_df = df_model[df_model['timestamp'] < time_70].copy()

val_df = df_model[
    (df_model['timestamp'] >= time_70) &
    (df_model['timestamp'] < time_85)
].copy()

test_df = df_model[df_model['timestamp'] >= time_85].copy()

# ------------------------------------------------
# 3. Prepare historical sequence inputs
# Order: oldest -> newest
# ------------------------------------------------

sequence_columns = ['lag_3', 'lag_2', 'lag_1']

def make_sequences(data):
    return np.log1p(
        data[sequence_columns].to_numpy(dtype=np.float32)
    )

X_seq_train_raw = make_sequences(train_df)
X_seq_val_raw = make_sequences(val_df)
X_seq_test_raw = make_sequences(test_df)

# Scale sequence values using training data only
sequence_scaler = StandardScaler()

sequence_scaler.fit(
    X_seq_train_raw.reshape(-1, 1)
)

def scale_sequences(values):
    scaled = sequence_scaler.transform(
        values.reshape(-1, 1)
    )
    return scaled.reshape(-1, 3, 1).astype(np.float32)

X_seq_train = scale_sequences(X_seq_train_raw)
X_seq_val = scale_sequences(X_seq_val_raw)
X_seq_test = scale_sequences(X_seq_test_raw)

# ------------------------------------------------
# 4. Prepare context numerical features
# ------------------------------------------------

def make_context_numeric(data):
    result = pd.DataFrame(index=data.index)

    result['log_memory'] = np.log1p(data['memoryMB'])
    result['log_history_count'] = np.log1p(data['history_count'])
    result['log_prev3_mean'] = np.log1p(data['prev3_mean'])
    result['log_prev3_std'] = np.log1p(data['prev3_std'])
    result['log_previous_gap'] = np.log1p(data['previous_gap_sec'])

    result['previous_hour_sin'] = data['previous_hour_sin']
    result['previous_hour_cos'] = data['previous_hour_cos']

    return result

X_num_train_raw = make_context_numeric(train_df)
X_num_val_raw = make_context_numeric(val_df)
X_num_test_raw = make_context_numeric(test_df)

context_scaler = StandardScaler()

X_num_train = context_scaler.fit_transform(
    X_num_train_raw
).astype(np.float32)

X_num_val = context_scaler.transform(
    X_num_val_raw
).astype(np.float32)

X_num_test = context_scaler.transform(
    X_num_test_raw
).astype(np.float32)

# ------------------------------------------------
# 5. Encode categorical features
# ------------------------------------------------

def make_categories(data):
    return pd.get_dummies(
        data[['runtime', 'region']].astype(str),
        columns=['runtime', 'region'],
        dtype=np.float32
    )

X_cat_train_df = make_categories(train_df)
X_cat_val_df = make_categories(val_df)
X_cat_test_df = make_categories(test_df)

# Align validation/test categories with training columns
X_cat_val_df = X_cat_val_df.reindex(
    columns=X_cat_train_df.columns,
    fill_value=0
)

X_cat_test_df = X_cat_test_df.reindex(
    columns=X_cat_train_df.columns,
    fill_value=0
)

X_cat_train = X_cat_train_df.to_numpy(dtype=np.float32)
X_cat_val = X_cat_val_df.to_numpy(dtype=np.float32)
X_cat_test = X_cat_test_df.to_numpy(dtype=np.float32)

# Combine numerical and categorical context
X_context_train = np.concatenate(
    [X_num_train, X_cat_train], axis=1
)

X_context_val = np.concatenate(
    [X_num_val, X_cat_val], axis=1
)

X_context_test = np.concatenate(
    [X_num_test, X_cat_test], axis=1
)

# ------------------------------------------------
# 6. Prepare target: next observed cold-start latency
# ------------------------------------------------

y_train = np.log1p(
    train_df['latency'].to_numpy(dtype=np.float32)
)

y_val = np.log1p(
    val_df['latency'].to_numpy(dtype=np.float32)
)

y_test = np.log1p(
    test_df['latency'].to_numpy(dtype=np.float32)
)

# ------------------------------------------------
# 7. Display final input dimensions
# ------------------------------------------------

print("SEQUENCE INPUTS")
print("Train:", X_seq_train.shape)
print("Validation:", X_seq_val.shape)
print("Test:", X_seq_test.shape)

print("\nCONTEXT INPUTS")
print("Train:", X_context_train.shape)
print("Validation:", X_context_val.shape)
print("Test:", X_context_test.shape)

print("\nTARGETS")
print("Train:", y_train.shape)
print("Validation:", y_val.shape)
print("Test:", y_test.shape)

print("\nCategorical features:", list(X_cat_train_df.columns))

print("\nMissing values check:")
print("Sequence:", np.isnan(X_seq_train).sum())
print("Context:", np.isnan(X_context_train).sum())
print("Target:", np.isnan(y_train).sum())

print("\nPreprocessing completed!")
```

```python

# PHASE 6B: ATTENTION-BASED BIDIRECTIONAL GRU

import tensorflow as tf
from tensorflow.keras import layers, Model
from tensorflow.keras.callbacks import (
    EarlyStopping,
    ReduceLROnPlateau,
    ModelCheckpoint
)

# 1. Check GPU
print("TensorFlow version:", tf.__version__)
print("GPU devices:", tf.config.list_physical_devices('GPU'))

# Enable GPU memory growth
for gpu in tf.config.list_physical_devices('GPU'):
    try:
        tf.config.experimental.set_memory_growth(gpu, True)
    except RuntimeError:
        pass

# 2. Define model inputs
sequence_input = layers.Input(
    shape=(3, 1),
    name='historical_sequence'
)

context_input = layers.Input(
    shape=(X_context_train.shape[1],),
    name='context_features'
)

# 3. Bidirectional GRU
gru_output = layers.Bidirectional(
    layers.GRU(
        64,
        return_sequences=True,
        dropout=0.2
    ),
    name='bidirectional_gru'
)(sequence_input)

# 4. Temporal attention mechanism
attention_scores = layers.Dense(
    1,
    activation='tanh',
    name='attention_score'
)(gru_output)

attention_weights = layers.Softmax(
    axis=1,
    name='attention_weights'
)(attention_scores)

weighted_sequence = layers.Multiply()(
    [gru_output, attention_weights]
)

attention_output = layers.Lambda(
    lambda x: tf.reduce_sum(x, axis=1),
    name='attention_context'
)(weighted_sequence)

# 5. Context branch
context_output = layers.Dense(
    32,
    activation='relu',
    name='context_dense'
)(context_input)

context_output = layers.Dropout(0.2)(context_output)

# 6. Feature fusion
combined = layers.Concatenate(
    name='feature_fusion'
)([attention_output, context_output])

combined = layers.Dense(
    64,
    activation='relu'
)(combined)

combined = layers.Dropout(0.2)(combined)

combined = layers.Dense(
    32,
    activation='relu'
)(combined)

# 7. Regression output
prediction = layers.Dense(
    1,
    activation='linear',
    name='latency_prediction'
)(combined)

# 8. Create model
model = Model(
    inputs=[sequence_input, context_input],
    outputs=prediction,
    name='ABiGRU_Cold_Start_Predictor'
)

# 9. Compile using RMSprop
model.compile(
    optimizer=tf.keras.optimizers.RMSprop(
        learning_rate=0.001
    ),
    loss=tf.keras.losses.Huber(),
    metrics=['mae']
)

# 10. Display architecture
model.summary()

# 11. Training callbacks
callbacks = [
    EarlyStopping(
        monitor='val_loss',
        patience=3,
        restore_best_weights=True,
        verbose=1
    ),
    ReduceLROnPlateau(
        monitor='val_loss',
        factor=0.5,
        patience=2,
        min_lr=0.00001,
        verbose=1
    ),
    ModelCheckpoint(
        '/content/best_abigru.keras',
        monitor='val_loss',
        save_best_only=True,
        verbose=1
    )
]

# 12. Train
history = model.fit(
    [X_seq_train, X_context_train],
    y_train,
    validation_data=(
        [X_seq_val, X_context_val],
        y_val
    ),
    epochs=10,
    batch_size=1024,
    callbacks=callbacks,
    verbose=1
)

print("\nABiGRU training completed!")
```

```python

import numpy as np
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Generate predictions on the untouched test set
pred_log = model.predict(
    [X_seq_test, X_context_test],
    batch_size=1024,
    verbose=1
).flatten()

# Convert log predictions and targets to original latency units
pred_latency = np.expm1(np.maximum(pred_log, 0))
actual_latency = np.expm1(y_test)

# Calculate evaluation metrics
mae = mean_absolute_error(actual_latency, pred_latency)
rmse = np.sqrt(mean_squared_error(actual_latency, pred_latency))
r2 = r2_score(actual_latency, pred_latency)

absolute_errors = np.abs(actual_latency - pred_latency)
p95_error = np.percentile(absolute_errors, 95)

print("\n========== ABiGRU TEST RESULTS ==========")
print(f"MAE:             {mae:.4f}")
print(f"RMSE:            {rmse:.4f}")
print(f"R2 Score:        {r2:.4f}")
print(f"P95 Abs Error:   {p95_error:.4f}")

print("\n========== SAMPLE PREDICTIONS ==========")
for i in range(10):
    print(
        f"Actual: {actual_latency[i]:.3f} | "
        f"Predicted: {pred_latency[i]:.3f} | "
        f"Error: {absolute_errors[i]:.3f}"
    )
```

```python

# PHASE 6C: CNN-LSTM WITH ATTENTION

import tensorflow as tf
from tensorflow.keras import layers, Model

# 1. Define model inputs
cnn_sequence_input = layers.Input(
    shape=(3, 1),
    name='historical_sequence'
)

cnn_context_input = layers.Input(
    shape=(X_context_train.shape[1],),
    name='context_features'
)

# 2. CNN feature extraction
cnn_output = layers.Conv1D(
    filters=64,
    kernel_size=2,
    padding='same',
    activation='relu',
    name='cnn_feature_extraction'
)(cnn_sequence_input)

# 3. LSTM layer
lstm_output = layers.LSTM(
    64,
    return_sequences=True,
    dropout=0.2,
    name='lstm_layer'
)(cnn_output)

# 4. Temporal attention mechanism
attention_scores = layers.Dense(
    1,
    activation='tanh',
    name='attention_score'
)(lstm_output)

attention_weights = layers.Softmax(
    axis=1,
    name='attention_weights'
)(attention_scores)

weighted_sequence = layers.Multiply()(
    [lstm_output, attention_weights]
)

attention_output = layers.Lambda(
    lambda x: tf.reduce_sum(x, axis=1),
    name='attention_context'
)(weighted_sequence)

# 5. Context branch
context_output = layers.Dense(
    32,
    activation='relu',
    name='context_dense'
)(cnn_context_input)

context_output = layers.Dropout(0.2)(context_output)

# 6. Feature fusion
combined = layers.Concatenate(
    name='feature_fusion'
)([attention_output, context_output])

combined = layers.Dense(
    64,
    activation='relu'
)(combined)

combined = layers.Dropout(0.2)(combined)

combined = layers.Dense(
    32,
    activation='relu'
)(combined)

# 7. Regression output
cnn_prediction = layers.Dense(
    1,
    activation='linear',
    name='latency_prediction'
)(combined)

# 8. Create model
cnn_lstm_model = Model(
    inputs=[cnn_sequence_input, cnn_context_input],
    outputs=cnn_prediction,
    name='CNN_LSTM_Cold_Start_Predictor'
)

# 9. Compile model
cnn_lstm_model.compile(
    optimizer=tf.keras.optimizers.RMSprop(
        learning_rate=0.001
    ),
    loss=tf.keras.losses.Huber(),
    metrics=['mae']
)

# 10. Display architecture
cnn_lstm_model.summary()

print("CNN-LSTM model created successfully!")
```

```python

import numpy as np
import pandas as pd

df_features = df_clean.copy()

df_features = df_features.sort_values(
    ["functionName", "timestamp"]
).reset_index(drop=True)

grouped = df_features.groupby("functionName", sort=False)

for lag in [1, 2, 3]:
    df_features[f"lag_{lag}"] = grouped["latency"].shift(lag)

df_features["history_count"] = grouped.cumcount()

df_features["prev3_mean"] = (
    grouped["latency"]
    .transform(lambda x: x.shift(1).rolling(3, min_periods=1).mean())
)

df_features["prev3_std"] = (
    grouped["latency"]
    .transform(lambda x: x.shift(1).rolling(3, min_periods=2).std())
)

df_features["prev_timestamp"] = grouped["timestamp"].shift(1)

df_features["time_gap_sec"] = (
    df_features["timestamp"] - df_features["prev_timestamp"]
).dt.total_seconds()

df_features["hour"] = df_features["timestamp"].dt.hour

df_features["hour_sin"] = np.sin(
    2 * np.pi * df_features["hour"] / 24
)

df_features["hour_cos"] = np.cos(
    2 * np.pi * df_features["hour"] / 24
)

df_features["log_memory"] = np.log1p(df_features["memoryMB"])

df_features["log_time_gap"] = np.log1p(
    df_features["time_gap_sec"].clip(lower=0)
)

print("Feature dataset shape:", df_features.shape)
print("\nNew features:")
print(df_features.columns.tolist())
```

```python

import numpy as np
import pandas as pd

group_cols = ['region', 'functionName']

df_model = df_clean.sort_values(
    group_cols + ['timestamp']
).reset_index(drop=True)

grouped = df_model.groupby(group_cols, sort=False)

df_model['lag_1'] = grouped['latency'].shift(1)
df_model['lag_2'] = grouped['latency'].shift(2)
df_model['lag_3'] = grouped['latency'].shift(3)

df_model['history_count'] = grouped.cumcount()

df_model['prev3_mean'] = (
    df_model.groupby(group_cols)['lag_1']
    .transform(lambda x: x.rolling(3, min_periods=1).mean())
)

df_model['prev3_std'] = (
    df_model.groupby(group_cols)['lag_1']
    .transform(lambda x: x.rolling(3, min_periods=2).std())
)

df_model['time_gap_sec'] = (
    df_model.groupby(group_cols)['timestamp']
    .diff()
    .dt.total_seconds()
)

hour_float = (
    df_model['timestamp'].dt.hour +
    df_model['timestamp'].dt.minute / 60
)

df_model['hour_sin'] = np.sin(
    2 * np.pi * hour_float / 24
)

df_model['hour_cos'] = np.cos(
    2 * np.pi * hour_float / 24
)

df_model = df_model[
    df_model['history_count'] > 0
].copy()

df_model['lag_2'] = df_model['lag_2'].fillna(
    df_model['lag_1']
)

df_model['lag_3'] = df_model['lag_3'].fillna(
    df_model['lag_1']
)

df_model['prev3_std'] = df_model['prev3_std'].fillna(0)

df_model['log_time_gap'] = np.log1p(
    df_model['time_gap_sec'].clip(lower=0)
)

df_model['log_memory'] = np.log1p(df_model['memoryMB'])

df_model = df_model.reset_index(drop=True)

print("Original feature engineering restored!")
print("Shape:", df_model.shape)
print("Unique functions:", df_model['functionName'].nunique())
print("Unique region-function groups:",
      df_model.groupby(group_cols).ngroups)

print("\nMissing engineered values:")
print(df_model[
    ['lag_1', 'lag_2', 'lag_3', 'prev3_mean',
     'prev3_std', 'time_gap_sec']
].isnull().sum())

df_model.to_csv(
    "/content/drive/MyDrive/FaaSNet_Project/engineered_dataset.csv",
    index=False
)

print("\nSaved to Google Drive!")
```

```python

# Identify context features
numeric_context = [
    "lag_1",
    "lag_2",
    "lag_3",
    "history_count",
    "prev3_mean",
    "prev3_std",
    "log_time_gap"
]

categorical_context = ["runtime", "region"]

# One-hot encode categorical features
X_context_all = pd.get_dummies(
    df_model[numeric_context + categorical_context],
    columns=categorical_context,
    dtype=float
)

print("Context feature shape:", X_context_all.shape)

# Historical sequence: previous 3 latency values
X_seq_all = df_model[
    ["lag_3", "lag_2", "lag_1"]
].to_numpy().reshape(-1, 3, 1)

# Target
y_all = np.log1p(df_model["latency"].to_numpy())

print("Sequence shape:", X_seq_all.shape)
print("Target shape:", y_all.shape)

print("\nContext feature count:", X_context_all.shape[1])
```

```python

import tensorflow as tf
from tensorflow.keras.layers import (
    Input, Conv1D, LSTM, Dense, Dropout,
    Softmax, Multiply, Lambda, Concatenate
)
from tensorflow.keras.models import Model
from tensorflow.keras.optimizers import RMSprop
from tensorflow.keras.losses import Huber

# Sequence input
sequence_input = Input(
    shape=(3, 1),
    name="historical_sequence"
)

# Context input
context_input = Input(
    shape=(24,),
    name="context_features"
)

# CNN feature extraction
x = Conv1D(
    filters=64,
    kernel_size=2,
    padding="same",
    activation="relu"
)(sequence_input)

# LSTM temporal learning
x = LSTM(
    64,
    return_sequences=True,
    dropout=0.2
)(x)

# Attention mechanism
attention_scores = Dense(
    1,
    activation="tanh"
)(x)

attention_weights = Softmax(axis=1)(
    attention_scores
)

weighted_sequence = Multiply()([
    x,
    attention_weights
])

attention_output = Lambda(
    lambda x: tf.reduce_sum(x, axis=1)
)(weighted_sequence)

# Context branch
context_output = Dense(
    32,
    activation="relu"
)(context_input)

context_output = Dropout(0.2)(context_output)

# Feature fusion
combined = Concatenate()([
    attention_output,
    context_output
])

combined = Dense(64, activation="relu")(combined)
combined = Dropout(0.2)(combined)

combined = Dense(32, activation="relu")(combined)

# Regression output
prediction = Dense(
    1,
    activation="linear"
)(combined)

# Create model
cnn_lstm_model = Model(
    inputs=[sequence_input, context_input],
    outputs=prediction,
    name="CNN_LSTM_Attention"
)

# Compile
cnn_lstm_model.compile(
    optimizer=RMSprop(learning_rate=0.001),
    loss=Huber(),
    metrics=["mae"]
)

cnn_lstm_model.summary()

print("CNN-LSTM model created successfully!")
```

```python

from tensorflow.keras.callbacks import (
    EarlyStopping,
    ReduceLROnPlateau,
    ModelCheckpoint
)

cnn_lstm_checkpoint = "/content/best_cnn_lstm.keras"

cnn_lstm_callbacks = [
    EarlyStopping(
        monitor="val_loss",
        patience=3,
        restore_best_weights=True,
        verbose=1
    ),

    ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.5,
        patience=2,
        min_lr=0.00001,
        verbose=1
    ),

    ModelCheckpoint(
        cnn_lstm_checkpoint,
        monitor="val_loss",
        save_best_only=True,
        verbose=1
    )
]

print("Callbacks configured successfully.")
print("Checkpoint:", cnn_lstm_checkpoint)
```
